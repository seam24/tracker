```python
# ------------------------------------------------------------
# Data from the RQ2 detailed regional table
# Values are percentages within each region.
# np.nan denotes NA, i.e., no mutants of that pattern were found in that region.
# ------------------------------------------------------------

regions = ["r1", "r2", "r3", "r4"]

data = {
    "black": {
        "Stable S": [0.4, 1.8, 2.0, 1.9],
        "Stable K": [88.3, 91.6, 91.7, 97.8],
        "S→K": [0.0, 0.6, 0.4, 0.2],
        "K→S": [0.1, 0.0, 0.0, 0.0],
        "S→K→S": [0.0, 0.0, 0.0, 0.0],
        "K→S→K": [10.5, 5.8, 5.3, 0.0],
        "Recur.": [0.7, 0.2, 0.6, 0.0],
    },
    "cookiecutter": {
        "Stable S": [30.0, 31.6, 5.9, 21.5],
        "Stable K": [70.0, 68.4, 88.2, 78.5],
        "S→K": [0.0, 0.0, 5.9, 0.0],
        "K→S": [0.0, 0.0, 0.0, 0.0],
        "S→K→S": [0.0, 0.0, 0.0, 0.0],
        "K→S→K": [0.0, 0.0, 0.0, 0.0],
        "Recur.": [0.0, 0.0, 0.0, 0.0],
    },
    "fastapi": {
        "Stable S": [86.8, 48.8, 44.0, 57.4],
        "Stable K": [13.2, 47.9, 54.5, 40.5],
        "S→K": [0.0, 3.3, 1.6, 1.8],
        "K→S": [0.0, 0.0, 0.0, 0.2],
        "S→K→S": [0.0, 0.0, 0.0, 0.0],
        "K→S→K": [0.0, 0.0, 0.0, 0.0],
        "Recur.": [0.0, 0.0, 0.0, 0.0],
    },
    "httpie": {
        "Stable S": [19.5, 27.2, 11.4, 41.0],
        "Stable K": [42.4, 37.1, 87.1, 52.0],
        "S→K": [38.1, 35.8, 1.4, 7.0],
        "K→S": [0.0, 0.0, 0.0, 0.0],
        "S→K→S": [0.0, 0.0, 0.0, 0.0],
        "K→S→K": [0.0, 0.0, 0.0, 0.0],
        "Recur.": [0.0, 0.0, 0.0, 0.0],
    },
    "sanic": {
        "Stable S": [0.0, np.nan, 0.0, np.nan],
        "Stable K": [100.0, np.nan, 76.3, np.nan],
        "S→K": [0.0, np.nan, 5.3, np.nan],
        "K→S": [0.0, np.nan, 1.3, np.nan],
        "S→K→S": [0.0, np.nan, 3.9, np.nan],
        "K→S→K": [0.0, np.nan, 0.0, np.nan],
        "Recur.": [0.0, np.nan, 13.2, np.nan],
    },
    "thefuck": {
        "Stable S": [30.4, 61.1, 41.4, 54.4],
        "Stable K": [57.6, 34.6, 53.4, 38.4],
        "S→K": [4.9, 1.4, 1.4, 2.2],
        "K→S": [6.3, 2.4, 2.1, 0.9],
        "S→K→S": [0.6, 0.5, 0.9, 3.6],
        "K→S→K": [0.0, 0.0, 0.5, 0.1],
        "Recur.": [0.2, 0.0, 0.3, 0.4],
    },
    "tqdm": {
        "Stable S": [43.8, 15.9, 18.8, 27.4],
        "Stable K": [31.1, 42.1, 43.7, 42.9],
        "S→K": [4.3, 11.6, 4.1, 4.5],
        "K→S": [1.0, 0.8, 3.6, 7.1],
        "S→K→S": [1.6, 2.1, 5.8, 1.5],
        "K→S→K": [2.3, 0.9, 0.2, 1.6],
        "Recur.": [15.8, 26.6, 23.7, 14.9],
    },
    "youtube-dl": {
        "Stable S": [74.1, 70.0, 77.7, np.nan],
        "Stable K": [17.9, 28.2, 20.0, np.nan],
        "S→K": [1.2, 0.4, 0.1, np.nan],
        "K→S": [2.0, 1.5, 0.0, np.nan],
        "S→K→S": [4.8, 0.0, 2.2, np.nan],
        "K→S→K": [0.0, 0.0, 0.0, np.nan],
        "Recur.": [0.0, 0.0, 0.0, np.nan],
    },
}

# ------------------------------------------------------------
# Dominant patterns based on RQ1
# ------------------------------------------------------------

dominant_stable = {
    "black": "Stable K",
    "cookiecutter": "Stable K",
    "fastapi": "Stable S",
    "httpie": "Stable K",
    "sanic": "Stable K",
    "thefuck": "Stable S",
    "tqdm": "Stable K",
    "youtube-dl": "Stable S",
}

dominant_nonstable = {
    "black": "K→S→K",
    "cookiecutter": "S→K",
    "fastapi": "S→K",
    "httpie": "S→K",
    "sanic": "Recur.",
    "thefuck": "S→K→S",
    "tqdm": "Recur.",
    "youtube-dl": "S→K→S",
}

projects = [
    "black",
    "cookiecutter",
    "fastapi",
    "httpie",
    "sanic",
    "thefuck",
    "tqdm",
    "youtube-dl",
]

# ------------------------------------------------------------
# Build plot rows: two rows per project
# ------------------------------------------------------------

rows = []
for project in projects:
    rows.append((project, dominant_stable[project], "Dominant stable"))
    rows.append((project, dominant_nonstable[project], "Dominant non-stable"))

n_rows = len(rows)
y_base = np.arange(n_rows)

# Four side-by-side bars per row
bar_height = 0.16
offsets = np.array([-1.5, -0.5, 0.5, 1.5]) * bar_height

fig_height = 0.48 * n_rows + 1.6
fig, ax = plt.subplots(figsize=(8, fig_height))

for i, region in enumerate(regions):
    values = []
    for project, pattern, _ in rows:
        values.append(data[project][pattern][i])

    y = y_base + offsets[i]
    ax.barh(
        y,
        values,
        height=bar_height,
        label=region,
    )

# ------------------------------------------------------------
# Axis formatting
# ------------------------------------------------------------

# Pattern labels on the y-axis
ytick_labels = []
for project, pattern, row_type in rows:
    if row_type == "Dominant stable":
        ytick_labels.append(f"{pattern}\n(stable)")
    else:
        ytick_labels.append(f"{pattern}\n(non-stable)")

ax.set_yticks(y_base)
ax.set_yticklabels(ytick_labels, fontsize=9)

ax.set_xlabel("Percentage within region (%)", fontsize=11)
ax.set_xlim(0, 105)
ax.grid(axis="x", linestyle="--", alpha=0.5)
ax.set_axisbelow(True)

# Put first project at the top
ax.invert_yaxis()

# Project labels on the far left, centered across the two rows
for p_idx, project in enumerate(projects):
    y_center = 2 * p_idx + 0.5
    ax.text(
        -0.18,
        y_center,
        project,
        transform=ax.get_yaxis_transform(),
        ha="right",
        va="center",
        fontsize=10,
        fontweight="bold",
        clip_on=False,
    )

# Optional separators between projects
for p_idx in range(1, len(projects)):
    ax.axhline(2 * p_idx - 0.5, color="0.85", linewidth=0.8)

ax.legend(
    title="Region",
    ncol=4,
    loc="lower center",
    bbox_to_anchor=(-0.12, -0.08),
    frameon=True,
    fontsize=8,
    title_fontsize=8
)

# Add value labels for reasonably visible bars
for container in ax.containers:
    for bar in container:
        width = bar.get_width()
        if np.isnan(width):
            continue
        if width >= 3.0:
            ax.text(
                width + 0.8,
                bar.get_y() + bar.get_height() / 2,
                f"{width:.1f}",
                va="center",
                ha="left",
                fontsize=8,
            )

plt.tight_layout()

# Save
plt.savefig("rq2_dominant_patterns_by_region.pdf", bbox_inches="tight")

plt.show()
```
